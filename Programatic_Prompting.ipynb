```python
import requests
import json


class OllamaClient:
    def __init__(self, base_url="http://localhost:11434", model="llama3.1"):
      self.base_url = base_url
      self.model = model

    def run(self, prompt: str):
      api_end_point = f"{self.base_url}/api/generate"

      headers = {"Content-Type": "application/json"}
      request_body = {"model": self.model,
                      "prompt": prompt,
                      "stream": False}

      response = requests.post(url = api_end_point, headers= headers,json=request_body)

      data = response.json()
      return data.get("response", "None")
```

# Examples Usage of Ollama Client

```python
llm = OllamaClient()
```

```python
result = llm.run("What is the Capital City of India?")
```
