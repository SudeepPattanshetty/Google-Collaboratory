# Call ollama model using python

```python
import requests
import json

request_body = {"model": "llama3.1",
                "prompt": "tell me a joke on computer"}
headers = {"Content-Type": "application/json"}

api_end_point = "http://localhost:11434/api/generate"

response = requests.post(url=api_end_point,headers=headers,json=request_body,stream=True)
print(response.text)
```

```python
full_data = response.text
type(full_data)
```

```python
lines = full_data.split("\n")
lines = [json.loads(line) for line in lines if line]
print(len(lines))

lines[0]
```

```python
final_result = ""
for i,line in enumerate(lines):
  if 'response' in line:
    print(f"Response {i}: {line['response']}")
    final_result += line['response']
```
